```python
from ucimlrepo import fetch_ucirepo
import pandas as pd

# Fetch the dataset
heart_disease = fetch_ucirepo(id=45)
X = heart_disease.data.features
y = heart_disease.data.targets

# Check for missing values in the features
missing_values = X.isnull().sum()

# Display the number of missing values in each feature
print("Missing Values in Features:")
print(missing_values)


from sklearn.impute import SimpleImputer

imputer = SimpleImputer(strategy='mean')
X_imputed = pd.DataFrame(imputer.fit_transform(X), columns=X.columns)

# Verify that missing values are handled
print("\nMissing Values After Imputation:")
print(X_imputed.isnull().sum())
```

```python
from sklearn.impute import SimpleImputer

# Check if there are missing values in categorical columns
categorical_features = X_imputed.select_dtypes(include=['object']).columns

if not categorical_features.empty:
    print("Categorical columns with missing values detected:")
    for col in categorical_features:
        if X_imputed[col].isnull().sum() > 0:
            print(f" - {col} has {X_imputed[col].isnull().sum()} missing values.")

    # Impute missing values in categorical columns with the mode
    imputer_cat = SimpleImputer(strategy='most_frequent')
    X_imputed[categorical_features] = imputer_cat.fit_transform(X_imputed[categorical_features])

    # Confirm that missing values are handled
    print("\nMissing Values After Imputation in Categorical Columns:")
    print(X_imputed[categorical_features].isnull().sum())
else:
    print("No categorical columns with missing values detected.")
```

```python
# Identify numeric features
numeric_features = X_imputed.select_dtypes(include=['number']).columns

# Check for outliers using boxplots (before handling outliers)
import matplotlib.pyplot as plt
import seaborn as sns

plt.figure(figsize=(5, 5))
sns.boxplot(data=X_imputed[numeric_features], palette='Set1')
plt.title("Boxplot of Numerical Features (Outliers Detection)")
plt.xticks(rotation=45, fontsize=12)
plt.tight_layout()
plt.show()
```

```python
from sklearn.preprocessing import StandardScaler

# Identify numerical features
numeric_features = X_imputed.select_dtypes(include=['int64', 'float64']).columns

# Apply scaling to numerical features
scaler = StandardScaler()
X_scaled = X_imputed.copy()
X_scaled[numeric_features] = scaler.fit_transform(X_imputed[numeric_features])

# Check the scaled features
print(X_scaled.head())
```

```python
from sklearn.preprocessing import OneHotEncoder

# One-hot encode the categorical features
encoder = OneHotEncoder(sparse_output=False, drop='first')  # Drop first to avoid multicollinearity
X_encoded = pd.DataFrame(encoder.fit_transform(X_imputed[categorical_features]), columns=encoder.get_feature_names_out())

# Merge the encoded features back with the original DataFrame (excluding original categorical columns)
X_final = pd.concat([X_scaled, X_encoded], axis=1)

# Display the final processed feature set
print(X_final.head())
```

```python
from sklearn.ensemble import RandomForestClassifier

# Train a RandomForest model to evaluate feature importance
model = RandomForestClassifier(random_state=42)
model.fit(X_final, y)

# Get and visualize feature importances
feature_importances = model.feature_importances_
sorted_idx = feature_importances.argsort()

# Plot the feature importances
plt.figure(figsize=(5, 5))
plt.barh(X_final.columns[sorted_idx], feature_importances[sorted_idx])
plt.title('Feature Importances')
plt.xlabel('Importance')
plt.ylabel('Feature')
plt.show()
```

```python
#Q3
import numpy as np
from sklearn.model_selection import train_test_split, GridSearchCV
from sklearn.linear_model import LogisticRegression
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import RandomForestClassifier
from sklearn.svm import SVC
from sklearn.metrics import classification_report
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense
from tensorflow.keras.utils import to_categorical


# Split the dataset into training and testing sets
X_train, X_test, y_train, y_test = train_test_split(X_final, y, test_size=0.2, random_state=42)

# Define models
models = {
    'Logistic Regression': LogisticRegression(max_iter=1000, random_state=42),
    'Decision Tree': DecisionTreeClassifier(random_state=42),
    'Random Forest': RandomForestClassifier(random_state=42),
    'SVM': SVC(probability=True, random_state=42)
}

# Hyperparameter tuning for Logistic Regression using GridSearchCV
logreg_params = {'C': [0.01, 0.1, 1, 10], 'solver': ['liblinear', 'saga']}
logreg_grid = GridSearchCV(LogisticRegression(max_iter=1000, random_state=42), logreg_params, cv=5)
logreg_grid.fit(X_train, y_train)
print('Best Logistic Regression Parameters:', logreg_grid.best_params_)

# Logistic Regression with the best parameters
models['Logistic Regression'] = logreg_grid.best_estimator_

# Hyperparameter tuning for Random Forest (example)
rf_params = {'n_estimators': [50, 100, 200], 'max_depth': [None, 10, 20]}
rf_grid = GridSearchCV(RandomForestClassifier(random_state=42), rf_params, cv=5)
rf_grid.fit(X_train, y_train)
print('Best Random Forest Parameters:', rf_grid.best_params_)

# Random Forest with the best parameters
models['Random Forest'] = rf_grid.best_estimator_

# Train and evaluate models
results = {}
for name, model in models.items():
    print(f"\n{name} Model:")
    model.fit(X_train, y_train)
    y_pred = model.predict(X_test)
    accuracy = model.score(X_test, y_test)
    results[name] = accuracy
    print(f"Accuracy: {accuracy:.2f}")
    print(classification_report(y_test, y_pred))

# Basic Neural Network using Keras (optional)
if len(np.unique(y)) > 2:  # Check if it's a multi-class problem
    y_train_nn = to_categorical(y_train)
    y_test_nn = to_categorical(y_test)
    output_units = len(np.unique(y))
    loss_function = 'categorical_crossentropy'
else:
    y_train_nn = y_train
    y_test_nn = y_test
    output_units = 1
    loss_function = 'binary_crossentropy'

# Define the neural network
nn = Sequential([
    Dense(64, activation='relu', input_dim=X_train.shape[1]),
    Dense(32, activation='relu'),
    Dense(output_units, activation='softmax' if output_units > 1 else 'sigmoid')
])

# Compile the model
nn.compile(optimizer='adam', loss=loss_function, metrics=['accuracy'])

# Train the model
history = nn.fit(X_train, y_train_nn, epochs=20, batch_size=32, validation_split=0.2, verbose=0)

# Evaluate the model
nn_loss, nn_accuracy = nn.evaluate(X_test, y_test_nn, verbose=0)
results['Neural Network'] = nn_accuracy

print("\nNeural Network Results:")
print(f"Accuracy: {nn_accuracy:.2f}")
```

```python
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score, roc_auc_score, confusion_matrix, roc_curve
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np
from sklearn.preprocessing import label_binarize

# Initialize dictionary to store results
evaluation_results = {}

# Function to evaluate a model
def evaluate_model(model, X_test, y_test, model_name):
    y_pred = model.predict(X_test)
    y_pred_proba = (
        model.predict_proba(X_test) if hasattr(model, "predict_proba") else 
        model.decision_function(X_test) if hasattr(model, "decision_function") else None
    )

    # Calculate metrics
    accuracy = accuracy_score(y_test, y_pred)
    precision = precision_score(y_test, y_pred, average='weighted')
    recall = recall_score(y_test, y_pred, average='weighted')
    f1 = f1_score(y_test, y_pred, average='weighted')
    
    # Adjust roc_auc_score for multiclass or binary case
    if y_pred_proba is not None:
        if len(np.unique(y_test)) > 2:  # Multiclass case
            roc_auc = roc_auc_score(y_test, y_pred_proba, multi_class="ovr", average="weighted")
        else:  # Binary case
            roc_auc = roc_auc_score(y_test, y_pred_proba[:, 1])
    else:
        roc_auc = None

    # Store metrics in the results dictionary
    evaluation_results[model_name] = {
        "Accuracy": accuracy,
        "Precision": precision,
        "Recall": recall,
        "F1-Score": f1,
        "ROC-AUC": roc_auc,
    }

    # Print metrics
    print(f"\n{model_name} Evaluation Metrics:")
    print(f"Accuracy: {accuracy:.2f}")
    print(f"Precision: {precision:.2f}")
    print(f"Recall: {recall:.2f}")
    print(f"F1-Score: {f1:.2f}")
    if roc_auc is not None:
        print(f"ROC-AUC: {roc_auc:.2f}")

    # Confusion Matrix
    cm = confusion_matrix(y_test, y_pred)
    plt.figure(figsize=(6, 4))
    sns.heatmap(cm, annot=True, fmt="d", cmap="Oranges", xticklabels=np.unique(y_test), yticklabels=np.unique(y_test))
    plt.title(f"{model_name} Confusion Matrix")
    plt.xlabel("Predicted Label")
    plt.ylabel("True Label")
    plt.show()

    # ROC Curve (if binary or probabilities available)
    if y_pred_proba is not None:
        plt.figure(figsize=(8, 6))
        
        if len(np.unique(y_test)) == 2:  # Binary classification
            fpr, tpr, thresholds = roc_curve(y_test, y_pred_proba[:, 1])
            plt.plot(fpr, tpr, label=f"{model_name} (AUC = {roc_auc:.2f})")
        else:  # Multi-class classification
            # One-vs-Rest for multi-class
            y_test_bin = label_binarize(y_test, classes=np.unique(y_test))  # Binarize the labels for multi-class
            for i in range(y_test_bin.shape[1]):
                fpr, tpr, thresholds = roc_curve(y_test_bin[:, i], y_pred_proba[:, i])
                plt.plot(fpr, tpr, label=f"Class {i} (AUC = {roc_auc:.2f})")
        
        # Diagonal line for random model (no discrimination)
        plt.plot([0, 1], [0, 1], "k--")
        plt.xlabel("False Positive Rate")
        plt.ylabel("True Positive Rate")
        plt.title(f"{model_name} ROC Curve")
        plt.legend(loc="lower right")
        plt.show()

# Example: Evaluate all models (assuming `models`, `X_test`, and `y_test` are already defined)
for name, model in models.items():
    evaluate_model(model, X_test, y_test, name)

# The function now handles ROC plotting for both binary and multi-class cases
```
